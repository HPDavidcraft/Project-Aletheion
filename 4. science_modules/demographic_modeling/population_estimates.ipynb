# Settlement population capacity estimates

This notebook applies the versioned capacity model to curator-supplied settlement area, inhabited fraction, housing density, household-size, and carrying-capacity ranges. It has no historical defaults and does not infer a census count. Replace the `None` / empty input values below with evidence-backed parameters and citations before running an estimate. Preserve this notebook version and parameter-set UUID with any writeback.

```python
from datetime import datetime, timezone
from decimal import Decimal
from uuid import UUID

from demographic_modeling.model import (
    DemographicInputs,
    DemographicRunProvenance,
    NumericRange,
    SettlementPopulationModel,
)

NOTEBOOK_VERSION = "1.0.0"

def run_estimate(
    *,
    site_id: UUID,
    settlement_area_source_id: UUID,
    settlement_area_hectares: tuple[Decimal, Decimal],
    inhabited_fraction: tuple[Decimal, Decimal],
    housing_density_dwellings_per_hectare: tuple[Decimal, Decimal],
    household_size_persons_per_dwelling: tuple[Decimal, Decimal],
    carrying_capacity_persons: tuple[Decimal, Decimal],
    parameter_set_id: UUID,
    bibliography_reference_ids: tuple[UUID, ...],
):
    inputs = DemographicInputs(
        site_id=site_id,
        settlement_area_source_id=settlement_area_source_id,
        settlement_area_hectares=NumericRange(*settlement_area_hectares),
        inhabited_fraction=NumericRange(*inhabited_fraction),
        housing_density_dwellings_per_hectare=NumericRange(
            *housing_density_dwellings_per_hectare
        ),
        household_size_persons_per_dwelling=NumericRange(
            *household_size_persons_per_dwelling
        ),
        carrying_capacity_persons=NumericRange(*carrying_capacity_persons),
    )
    provenance = DemographicRunProvenance(
        parameter_set_id=parameter_set_id,
        notebook_version=NOTEBOOK_VERSION,
        execution_timestamp=datetime.now(timezone.utc),
        bibliography_reference_ids=bibliography_reference_ids,
    )
    return SettlementPopulationModel().estimate(inputs, provenance)
```

```python
# Populate each range from reviewed evidence; values are intentionally not fabricated.
site_id = None
settlement_area_source_id = None
settlement_area_hectares = None
inhabited_fraction = None
housing_density_dwellings_per_hectare = None
household_size_persons_per_dwelling = None
carrying_capacity_persons = None
parameter_set_id = None
bibliography_reference_ids = ()

required_inputs = (
    site_id,
    settlement_area_source_id,
    settlement_area_hectares,
    inhabited_fraction,
    housing_density_dwellings_per_hectare,
    household_size_persons_per_dwelling,
    carrying_capacity_persons,
    parameter_set_id,
)
if any(value is None for value in required_inputs) or not bibliography_reference_ids:
    print("Provide evidence-backed input ranges, UUIDs, and bibliography references before estimating.")
else:
    estimate = run_estimate(
        site_id=site_id,
        settlement_area_source_id=settlement_area_source_id,
        settlement_area_hectares=settlement_area_hectares,
        inhabited_fraction=inhabited_fraction,
        housing_density_dwellings_per_hectare=housing_density_dwellings_per_hectare,
        household_size_persons_per_dwelling=household_size_persons_per_dwelling,
        carrying_capacity_persons=carrying_capacity_persons,
        parameter_set_id=parameter_set_id,
        bibliography_reference_ids=bibliography_reference_ids,
    )
    print(estimate.to_json())
```
