# Stable carbon and nitrogen isotope analysis

This R-kernel template validates an authorized CSV of δ13C and δ15N measurements, summarizes the supplied groups, and creates a biplot with reported uncertainty intervals. It does not infer a diet. Add only an analyst-authored interpretation grounded in cited reference data and the recorded laboratory method. Treat plot and text as computational outputs: register them as Media_Assets through the authorized writeback path.

```r
source("isotope_analysis.R")

input_csv <- Sys.getenv("PLEROMA_CN_INPUT", unset = "")
output_dir <- Sys.getenv("PLEROMA_OUTPUT_DIR", unset = "")
parameter_set_id <- Sys.getenv("PLEROMA_PARAMETER_SET_ID", unset = "")
notebook_revision <- Sys.getenv("PLEROMA_NOTEBOOK_REVISION", unset = "")
dietary_interpretation <- ""

if (!nzchar(input_csv) || !nzchar(output_dir) ||
    !nzchar(parameter_set_id) || !nzchar(notebook_revision)) {
  stop("Set the authorized input, output directory, parameter-set UUID, and notebook revision")
}
if (!nzchar(trimws(dietary_interpretation))) {
  stop("Enter an analyst-authored, cited interpretation before producing outputs")
}
```

```r
analysis <- run_cn_analysis(
  input_csv = input_csv,
  output_dir = output_dir,
  parameter_set_id = parameter_set_id,
  notebook_revision = notebook_revision,
  dietary_interpretation = dietary_interpretation
)
analysis$summary
analysis$outputs
```

The generated summary describes measurements only. Keep the source CSV and its checksum, reference citations, uncertainty basis, parameter-set UUID, notebook revision, interpretation, and registered output Media_Asset identifiers together in the analysis record.
